```python
from dotenv import load_dotenv
from google import genai
import os

load_dotenv(".env.example")

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
def validate_topic(topic):
    if not topic.strip():
        print("Please enter a topic.")
        return False

    return True
```

```python
def search_papers(topic):

    if not validate_topic(topic):
        return

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=f"""
You are a research assistant.

Suggest 3 real and influential research papers about {topic}.

For each paper provide:
- Title
- Authors
- Publication Year
- Research Area
- Short Summary (2-3 sentences)
"""
    )

    print(response.text)
```

```python
search_papers("machine learning")
```

```python
def summarize_paper(title):

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=f"""
Provide a concise summary of the research paper:

{title}

Include:
- Main idea
- Key contributions
- Importance of the paper
"""
    )

    print(response.text)
```

```python
summarize_paper("Attention Is All You Need")
```

```python
def select_tool(user_request):

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=f"""
You are a tool selection assistant.

Available tools:
1. search_papers - use when the user wants to find research papers.
2. summarize_paper - use when the user wants a summary of a specific paper.

User request:
{user_request}

Return ONLY one of:
search_papers
summarize_paper
"""
    )

    selected_tool = response.text.strip()

    print("Selected tool:", selected_tool)

    if selected_tool == "search_papers":
        search_papers(user_request)

    elif selected_tool == "summarize_paper":
        summarize_paper(user_request)
```

```python
select_tool("Find papers about machine learning")
```

```python
select_tool("Give me a summary of Attention Is All You Need")
```
